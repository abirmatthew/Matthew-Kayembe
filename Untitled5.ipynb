```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 5000

data = pd.DataFrame({

    'Users_Connected':
        np.random.randint(50, 1000, n),

    'Download_Throughput':
        np.random.uniform(10, 500, n),

    'Upload_Throughput':
        np.random.uniform(5, 100, n),

    'Latency':
        np.random.uniform(5, 100, n),

    'Packet_Loss':
        np.random.uniform(0, 5, n),

    'CPU_Usage':
        np.random.uniform(10, 95, n),

    'Cell_Occupancy':
        np.random.uniform(20, 100, n)
})

data['Congestion_Level'] = (
    0.3 * data['Users_Connected']
    + 5 * data['Latency']
    + 15 * data['Packet_Loss']
    + 3 * data['Cell_Occupancy']
) / 100

data.to_csv('telecom_dataset.csv', index=False)

print(data.head())
```

```python
import pandas as pd

data = pd.read_csv('telecom_dataset.csv')

data.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

X = data.drop('Congestion_Level', axis=1)

y = data['Congestion_Level']

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
```

```python
import numpy as np

time_steps = 10

X_lstm = []
y_lstm = []

for i in range(time_steps, len(X_scaled)):
    X_lstm.append(X_scaled[i-time_steps:i])
    y_lstm.append(y.iloc[i])

X_lstm = np.array(X_lstm)
y_lstm = np.array(y_lstm)

print(X_lstm.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_lstm,
    y_lstm,
    test_size=0.2,
    random_state=42
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout

model = Sequential()

model.add(
    LSTM(
        128,
        return_sequences=True,
        input_shape=(10,7)
    )
)

model.add(Dropout(0.2))

model.add(
    LSTM(
        64
    )
)

model.add(Dropout(0.2))

model.add(
    Dense(
        32,
        activation='relu'
    )
)

model.add(
    Dense(
        1
    )
)
```

```python
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

model.summary()
```

```python
history = model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=64
)
```

```python
loss, mae = model.evaluate(
    X_test,
    y_test
)

print("MAE =", mae)
```

```python
prediction = model.predict(X_test)

print(prediction[:10])
```

```python
model.save(
    'telecom_congestion_lstm.h5'
)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
import numpy as np

predictions = model.predict(X_test)

mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print("MSE :", mse)
print("RMSE :", rmse)
print("MAE :", mae)
print("R² :", r2)
```

```python
y_test_class = (y_test >= 5).astype(int)
```

```python
predictions = model.predict(X_test)

y_pred_class = (predictions >= 5).astype(int)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test_class,
    y_pred_class
)

print("Accuracy :", accuracy)
```

```python
from sklearn.metrics import precision_score

precision = precision_score(
    y_test_class,
    y_pred_class
)

print("Precision :", precision)
```

```python
from sklearn.metrics import recall_score

recall = recall_score(
    y_test_class,
    y_pred_class
)

print("Recall :", recall)
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(
    y_test_class,
    y_pred_class
)

print("F1-Score :", f1)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test_class,
    y_pred_class
)

print(cm)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm
)

disp.plot()

plt.show()
```
